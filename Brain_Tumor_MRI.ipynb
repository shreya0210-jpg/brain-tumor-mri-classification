Potential future improvements include evaluating performance on independent test data, expanding the dataset, improving class-wise evaluation, tuning model hyperparameters, and deploying the application in a production-like environment.

Overall, this project demonstrates the application of CNN-based image classification and MobileNetV2 transfer learning to brain MRI analysis. It is intended for educational and experimental purposes and does not replace professional medical evaluation or clinically validated diagnostic systems.

# 1. Dataset Understanding

The dataset consists of brain MRI images belonging to four classes:
- Pituitary Tumor
- Meningioma Tumor
- No Tumor
- Glioma Tumor

The dataset is divided into training, validation, and testing sets. The dataset will be explored to understand class distribution, image dimensions, and representative MRI samples before model development.

````python
# Define dataset paths

train_path = "/content/train/train"
valid_path = "/content/valid/valid"
test_path = "/content/test/test"

# Define the four classes
class_names = ["pituitary", "meningioma", "no_tumor", "glioma"]

# Count images in each dataset
dataset_counts = {}

for dataset_name, dataset_path in [
    ("Training", train_path),
    ("Validation", valid_path),
    ("Test", test_path)
]:
    dataset_counts[dataset_name] = {}

    for class_name in class_names:
        class_path = os.path.join(dataset_path, class_name)

        count = len([
            file for file in os.listdir(class_path)
            if file.lower().endswith((".jpg", ".jpeg", ".png"))
        ])

        dataset_counts[dataset_name][class_name] = count

# Display summary
dataset_summary = pd.DataFrame(dataset_counts).T

print("Dataset Summary:")
display(dataset_summary)

print("\nTotal images in each dataset:")
print(dataset_summary.sum(axis=1))
````

````python
# Chart 1: Class Distribution in Training Dataset

plt.figure(figsize=(8, 5))

plt.bar(
    class_names,
    [dataset_counts["Training"][class_name] for class_name in class_names]
)

plt.title("Training Dataset - Class Distribution")
plt.xlabel("Brain Tumor Class")
plt.ylabel("Number of Images")

plt.show()
````

````python
# Check image dimensions in the training dataset

image_dimensions = []

for class_name in class_names:
    class_path = os.path.join(train_path, class_name)

    image_files = [
        file for file in os.listdir(class_path)
        if file.lower().endswith((".jpg", ".jpeg", ".png"))
    ]

    for file in image_files:
        image_path = os.path.join(class_path, file)

        with Image.open(image_path) as img:
            width, height = img.size
            image_dimensions.append((width, height))

# Convert to DataFrame
dimensions_df = pd.DataFrame(
    image_dimensions,
    columns=["Width", "Height"]
)

print("Image Dimension Summary:")
display(dimensions_df.describe())

print("\nUnique image dimensions:", dimensions_df.drop_duplicates().shape[0])
````

````python
# Calculate average pixel intensity for each class

class_pixel_means = {}

for class_name in class_names:
    class_path = os.path.join(train_path, class_name)

    image_files = [
        file for file in os.listdir(class_path)
        if file.lower().endswith((".jpg", ".jpeg", ".png"))
    ]

    pixel_means = []

    for file in image_files:
        image_path = os.path.join(class_path, file)

        image = Image.open(image_path).convert("L")
        image_array = np.array(image)

        pixel_means.append(image_array.mean())

    class_pixel_means[class_name] = np.mean(pixel_means)

# Plot average pixel intensity
plt.figure(figsize=(8, 5))

plt.bar(
    [name.replace("_", " ").title() for name in class_pixel_means.keys()],
    class_pixel_means.values()
)

plt.title("Average Pixel Intensity by Class")
plt.xlabel("Brain MRI Class")
plt.ylabel("Average Pixel Intensity")

plt.show()
````

### Pixel Intensity Analysis

Average pixel intensity was analyzed for each MRI class to understand differences in overall image brightness.

This visualization provides an additional perspective on the image data before preprocessing. Pixel intensity differences may result from variations in MRI image appearance and acquisition conditions.

The analysis is exploratory and does not indicate that pixel intensity alone can distinguish tumor categories.

# 2. Data Preprocessing

The MRI images will be resized to 224 × 224 pixels so that they can be used as input to the deep learning models. Pixel values will be normalized to the range 0–1 to make model training more stable.

Separate data generators will be created for the training, validation, and test datasets.

````python
# Image preprocessing settings

IMG_SIZE = (224, 224)
BATCH_SIZE = 32

# Training data generator
train_datagen = ImageDataGenerator(
    rescale=1./255
)

# Validation and test data generators
valid_datagen = ImageDataGenerator(
    rescale=1./255
)

test_datagen = ImageDataGenerator(
    rescale=1./255
)

# Load training images
train_generator = train_datagen.flow_from_directory(
    train_path,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    shuffle=True
)

# Load validation images
valid_generator = valid_datagen.flow_from_directory(
    valid_path,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    shuffle=False
)

# Load test images
test_generator = test_datagen.flow_from_directory(
    test_path,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    shuffle=False
)

print("\nClass indices:")
print(train_generator.class_indices)
````

# 3. Data Augmentation

Data augmentation is applied to the training images to increase the variety of training examples and help reduce overfitting.

The augmentation techniques used include rotation, horizontal flipping, zooming, and shifting. Augmentation is applied only to the training dataset, while validation and test images are only resized and normalized.

````python
# Data augmentation for training images

train_datagen_aug = ImageDataGenerator(
    rescale=1./255,
    rotation_range=15,
    width_shift_range=0.10,
    height_shift_range=0.10,
    zoom_range=0.10,
    horizontal_flip=True
)

# Create augmented training generator
train_generator_aug = train_datagen_aug.flow_from_directory(
    train_path,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    shuffle=True
)

print("Augmented training generator created successfully!")
````

### Augmentation Visualization

The augmented images demonstrate how transformations such as rotation, shifting, zooming, and horizontal flipping create variations of the original MRI images.

These transformations increase the diversity of the training data and can help the model generalize better to unseen images.

````python
# Display augmented MRI images

images, labels = next(train_generator_aug)

plt.figure(figsize=(12, 8))

for i in range(8):
    plt.subplot(2, 4, i + 1)

    plt.imshow(images[i])

    class_index = np.argmax(labels[i])
    class_name = list(train_generator_aug.class_indices.keys())[class_index]

    plt.title(class_name.replace("_", " ").title())
    plt.axis("off")

plt.suptitle("Augmented MRI Training Images", fontsize=16)
plt.tight_layout()
plt.show()
````

# 4. Custom CNN Model

A Convolutional Neural Network (CNN) will be developed from scratch for multi-class brain MRI classification.

The model uses convolutional layers to learn image features, batch normalization to stabilize training, max-pooling to reduce spatial dimensions, and dropout to reduce overfitting. A final softmax layer produces probabilities for the four MRI classes.

````python
# Build Custom CNN model

custom_cnn = keras.Sequential([

    layers.Input(shape=(224, 224, 3)),

    # Block 1
    layers.Conv2D(32, (3, 3), activation="relu"),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),

    # Block 2
    layers.Conv2D(64, (3, 3), activation="relu"),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),

    # Block 3
    layers.Conv2D(128, (3, 3), activation="relu"),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),

    # Classification layers
    layers.Flatten(),
    layers.Dense(128, activation="relu"),
    layers.BatchNormalization(),
    layers.Dropout(0.5),

    # Output layer
    layers.Dense(4, activation="softmax")
])

custom_cnn.summary()
````

````python
# Compile the Custom CNN

custom_cnn.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

print("Custom CNN compiled successfully!")
````

````python
# Lightweight Custom CNN

custom_cnn = keras.Sequential([

    layers.Input(shape=(224, 224, 3)),

    # Block 1
    layers.Conv2D(16, (3, 3), activation="relu"),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),

    # Block 2
    layers.Conv2D(32, (3, 3), activation="relu"),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),

    # Block 3
    layers.Conv2D(64, (3, 3), activation="relu"),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),

    # Classification
    layers.GlobalAveragePooling2D(),
    layers.Dense(64, activation="relu"),
    layers.BatchNormalization(),
    layers.Dropout(0.5),

    # 4 classes
    layers.Dense(4, activation="softmax")
])

custom_cnn.summary()
````

````python
# Compile the lightweight Custom CNN

custom_cnn.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

print("Custom CNN compiled successfully!")
````

````python
# Train the Custom CNN

EPOCHS = 8

early_stopping = keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=3,
    restore_best_weights=True
)

checkpoint = keras.callbacks.ModelCheckpoint(
    "best_custom_cnn.keras",
    monitor="val_loss",
    save_best_only=True
)

history_custom = custom_cnn.fit(
    train_generator_aug,
    validation_data=valid_generator,
    epochs=EPOCHS,
    callbacks=[early_stopping, checkpoint]
)
````

````python
# Plot Custom CNN training history

plt.figure(figsize=(8, 5))

plt.plot(history_custom.history["accuracy"], label="Training Accuracy")
plt.plot(history_custom.history["val_accuracy"], label="Validation Accuracy")

plt.title("Custom CNN - Training and Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)

plt.show()
````

````python
# Plot Custom CNN loss

plt.figure(figsize=(8, 5))

plt.plot(history_custom.history["loss"], label="Training Loss")
plt.plot(history_custom.history["val_loss"], label="Validation Loss")

plt.title("Custom CNN - Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)

plt.show()
````

````python
# Evaluate Custom CNN on the test dataset

test_loss, test_accuracy = custom_cnn.evaluate(test_generator, verbose=1)

print(f"\nCustom CNN Test Loss: {test_loss:.4f}")
print(f"Custom CNN Test Accuracy: {test_accuracy:.4f}")
print(f"Custom CNN Test Accuracy: {test_accuracy * 100:.2f}%")
````

````python
from sklearn.metrics import classification_report

# Generate predictions on the test dataset
test_generator.reset()

predictions = custom_cnn.predict(test_generator)

# Convert probabilities to predicted class indices
predicted_classes = np.argmax(predictions, axis=1)

# Get actual class indices
true_classes = test_generator.classes

# Class names in the correct Keras order
class_labels = list(test_generator.class_indices.keys())

# Classification report
report = classification_report(
    true_classes,
    predicted_classes,
    target_names=class_labels,
    digits=4
)

print("Custom CNN Classification Report:\n")
print(report)
````

````python
from sklearn.metrics import confusion_matrix

# Create confusion matrix
cm = confusion_matrix(true_classes, predicted_classes)

# Plot confusion matrix
plt.figure(figsize=(7, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=class_labels,
    yticklabels=class_labels
)

plt.title("Custom CNN - Confusion Matrix")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")

plt.show()
````

# 5. Transfer Learning with MobileNetV2

Transfer learning is used to take advantage of features learned by a pretrained model. MobileNetV2 pretrained on ImageNet will be used as the feature extractor, with a new classification layer added for the four brain MRI classes.

Initially, the pretrained layers will be frozen so that only the newly added classification layers are trained.

````python
# Build MobileNetV2 transfer learning model

base_model = tf.keras.applications.MobileNetV2(
    input_shape=(224, 224, 3),
    include_top=False,
    weights="imagenet"
)

# Freeze pretrained layers
base_model.trainable = False

# Build the complete model
mobilenet_model = keras.Sequential([
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.Dropout(0.3),
    layers.Dense(4, activation="softmax")
])

mobilenet_model.summary()
````

````python
# Compile MobileNetV2 model

mobilenet_model.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

print("MobileNetV2 model compiled successfully!")
````

````python
# Train MobileNetV2

EPOCHS = 8

early_stopping_mobilenet = keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=3,
    restore_best_weights=True
)

checkpoint_mobilenet = keras.callbacks.ModelCheckpoint(
    "best_mobilenetv2.keras",
    monitor="val_loss",
    save_best_only=True
)

history_mobilenet = mobilenet_model.fit(
    train_generator_aug,
    validation_data=valid_generator,
    epochs=EPOCHS,
    callbacks=[
        early_stopping_mobilenet,
        checkpoint_mobilenet
    ]
)
````

````python
plt.figure(figsize=(8, 5))

plt.plot(
    history_mobilenet.history["loss"],
    label="Training Loss"
)

plt.plot(
    history_mobilenet.history["val_loss"],
    label="Validation Loss"
)

plt.title("MobileNetV2 - Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)

plt.show()
````

````python
# Evaluate MobileNetV2 on the test dataset

test_generator.reset()

test_loss_mobilenet, test_accuracy_mobilenet = mobilenet_model.evaluate(
    test_generator,
    verbose=1
)

print(f"\nMobileNetV2 Test Loss: {test_loss_mobilenet:.4f}")
print(f"MobileNetV2 Test Accuracy: {test_accuracy_mobilenet:.4f}")
print(f"MobileNetV2 Test Accuracy: {test_accuracy_mobilenet * 100:.2f}%")
````

````python
from sklearn.metrics import classification_report

# Reset test generator
test_generator.reset()

# Generate predictions
mobilenet_predictions = mobilenet_model.predict(test_generator)

# Convert probabilities to class labels
mobilenet_predicted_classes = np.argmax(
    mobilenet_predictions,
    axis=1
)

# Actual labels
mobilenet_true_classes = test_generator.classes

# Class names in correct order
mobilenet_class_labels = list(
    test_generator.class_indices.keys()
)

# Classification report
mobilenet_report = classification_report(
    mobilenet_true_classes,
    mobilenet_predicted_classes,
    target_names=mobilenet_class_labels,
    digits=4
)

print("MobileNetV2 Classification Report:\n")
print(mobilenet_report)
````

````python
from sklearn.metrics import confusion_matrix

# Create confusion matrix
cm_mobilenet = confusion_matrix(
    mobilenet_true_classes,
    mobilenet_predicted_classes
)

# Plot confusion matrix
plt.figure(figsize=(7, 6))

sns.heatmap(
    cm_mobilenet,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=mobilenet_class_labels,
    yticklabels=mobilenet_class_labels
)

plt.title("MobileNetV2 - Confusion Matrix")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")

plt.show()
````

````python
# Compare Custom CNN and MobileNetV2

model_comparison = pd.DataFrame({
    "Model": [
        "Custom CNN",
        "MobileNetV2"
    ],
    "Test Accuracy": [
        test_accuracy,
        test_accuracy_mobilenet
    ]
})

model_comparison["Test Accuracy (%)"] = (
    model_comparison["Test Accuracy"] * 100
).round(2)

display(model_comparison)
````

````python
plt.figure(figsize=(8, 5))

plt.bar(
    model_comparison["Model"],
    model_comparison["Test Accuracy (%)"]
)

plt.title("Model Comparison - Test Accuracy")
plt.xlabel("Model")
plt.ylabel("Test Accuracy (%)")
plt.ylim(0, 100)

plt.show()
````

### Model Comparison Insights

The Custom CNN achieved a test accuracy of 32.52%, while the MobileNetV2 transfer-learning model achieved a test accuracy of 78.86%.

MobileNetV2 performed substantially better than the Custom CNN on the test dataset. The pretrained model benefits from features learned from a large image dataset, which can provide a stronger starting point for image classification than a small CNN trained entirely from scratch.

The results demonstrate the usefulness of transfer learning for this brain MRI image classification project. However, the model is developed for educational and project demonstration purposes and should not be considered a clinical diagnostic system.
